# 4 Completitud de las columnas clave

Revisamos qué % de filas tiene dato real en las columnas que nos interesan para el análisis. 

```python
columnas_interes = [
    'text', 'stars', 'rating', 'publishedAtDate', 'language',
    'originalLanguage', 'reviewDetailedRating/Comida', 'reviewDetailedRating/Servicio',
    'reviewDetailedRating/Ambiente', 'reviewContext/Tiempo de espera',
    'responseFromOwnerText', 'likesCount', 'reviewerNumberOfReviews',
    'isLocalGuide', 'title'
]

for c in columnas_interes: 
    if c in df.columns: 
        nn = df[c].notna().sum()
        print(f'{c:45s} -> no-nulos {nn}/{len(df)}')
    else: 
        print(f'{c:45s} -> columna no encontrada')
```
